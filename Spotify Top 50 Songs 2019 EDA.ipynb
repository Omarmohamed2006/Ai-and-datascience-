```python
df['Energy_to_Danceability_ratio']=df['Energy']/df['Danceability']
df['Energy_to_Danceability_ratio'].head(3)
```

```python
mean_pouplar=df['Popularity'].mean()
mean_bpm=df['Beats.Per.Minute'].mean()
df[(df['Popularity']<=mean_pouplar)& (df['Beats.Per.Minute']<=mean_bpm)]
```

```python
def get_tempo(bpm):
    if bpm<100:
        return 'slow'
    elif bpm<=140:
        return 'Medium'
    else:
        return 'Fast'


df['Tempo_class']=df['Beats.Per.Minute'].apply(get_tempo)
df['Tempo_class']
```

```python
df.groupby('Artist.Name').agg({
    'Popularity':'mean',
    'Energy':'max',
    'Track.Name':'count'
})
```

```python
plt.figure(figsize=(10, 5))
top_genres = df['Genre'].value_counts().head(7)
sns.barplot(x=top_genres.values, y=top_genres.index, palette='viridis')
plt.title('Top 7 Music Genres in Top 50', fontsize=14, fontweight='bold')
plt.xlabel('Number of Songs')
plt.ylabel('Genre')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.show()
```

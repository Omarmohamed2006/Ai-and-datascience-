"Winter"

"Spring"

"Summer"

"Autumn"

````python
month_num=int(input("please enter month number"))
if month_num==12 or month_num==1 or month_num==2:
    print("winter")
elif month_num==3 or month_num==4 or month_num==5:
    print("spring")
elif month_num==6 or month_num==7 or month_num==8:
    print("summer")
else:
    print("autumn")
````

Write a program that:

Takes an integer number from the user

Calculates its factorial using a for loop

Prints the result

Example:

Input → 5

Output → 120

````python
num=int(input("please enter a number"))
factorial=1
for i in range(1,num+1):
    factorial*=i
    
print(factorial)
````

Given a list of numbers:

if number is even → add it to even_list

otherwise → add it to odd_list
Print both lists at the end.

````python
list=[1,2,3,4,5,6]
even_list=[]
ood_list=[]
for i in list:
    if i%2==0:
        even_list.append(i)
    elif i%2!=0:
        ood_list.append(i)
    
    
print(even_list)
print(ood_list)
````

Level 4 – Advanced Thinking with for Loop

Write a program that:

Checks if a number is prime using a for loop

````python
num=int(input("please enter a number"))
if num<=1:
    print("not prime")
else:
    is_prime=True
    for i in range(2,num):
        if num%i==0:
            is_prime=False
            break
    
    
if is_prime:
            print("prime number")
else:
    print("not prime")
````
